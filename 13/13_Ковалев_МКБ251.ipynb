```python
import json
import requests
from os import getenv
from dotenv import load_dotenv
```

```python
# Загрузка переменных окружения из .env
load_dotenv()

VT_API_KEY = getenv("VT_API_KEY")

if not VT_API_KEY:
    raise ValueError("API ключ не найден! Задайте переменную окружения VT_API_KEY.")
```

#### **Этап 2. Авторизация и запрос**
- Напишите Python-скрипт, который выполняет авторизацию по API-ключу или другому методу. Если используется публичное API, настройте переменную окружения или передайте ключ в заголовках.
- Выполните запрос, возвращающий JSON-ответ. Например, получение статуса сканирования файла или правил фильтрации трафика.

```python
URL = f"https://www.virustotal.com/api/v3/"

headers = {
    "x-apikey": VT_API_KEY
}

try:
    while True:
        print("Выберите режим сканирования:\n"
            "1 - Скан хеша файла;\n"
            "2 - Скан IP-адреса;\n"
            "3 - Выход\n")
        MODE = int(input('Ввести и нажать Enter: '))
        if MODE in range(1, 4):
            break

    if MODE == 1:
        FILE_HASH = input('Скопируйте сюда хеш файла: ')
        if not FILE_HASH: raise ValueError('Укажите хеш файла для проверки!')
        
        response = requests.get(url=URL+f"files/{FILE_HASH}", headers=headers)

    elif MODE == 2:
        IP = input('Скопируйте сюда IP-адрес: ')
        if not IP: raise ValueError('Укажите IP-адрес для проверки!')

        response = requests.get(url=URL+f'ip_addresses/{IP}', headers=headers)

    elif MODE == 3:
        raise KeyboardInterrupt('Процесс прерван пользователем.')

    response.raise_for_status()
    data = response.json()
    print("Запрос выполнен успешно.")

except Exception as e:
    print(e)
```
